```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import linear_model
```

```python
# load the data

c1 = pd.read_csv('q3_1.csv')
c2 = pd.read_csv('q3_2.csv')
```

```python
# test train split

c1_train = c1[:int(0.8 * len(c1))]
c1_test = c1[int(0.8 * len(c1)):]
c2_train = c2[:int(0.8 * len(c2))]
c2_test = c2[int(0.8 * len(c2)):]
```

```python
X1_train = c1_train.drop(columns=["ChocolateType"])
y1_train = c1_train["ChocolateType"]
X1_test = c1_test.drop(columns="ChocolateType")
y1_test = c1_test["ChocolateType"]
X2_train = c2_train.drop(columns=["EnergyConsumption"])
y2_train = c2_train["EnergyConsumption"]
X2_test = c2_test.drop(columns=["EnergyConsumption"])
y2_test = c2_test["EnergyConsumption"]
```

```python
from sklearn import linear_model


logistic_model = linear_model.LogisticRegression()

# NAIVE LOGISTIC
logistic_model.fit(X1_train, y1_train)
naive_preds = logistic_model.predict(X1_test)
print((naive_preds == y1_test).mean())
```

```python
from sklearn import linear_model


logistic_model = linear_model.LogisticRegression()

# NAIVE LOGISTIC
logistic_model.fit(X1_train.drop(columns="SerialNo"), y1_train)
naive_preds = logistic_model.predict(X1_test.drop(columns="SerialNo"))
print((naive_preds == y1_test).mean())
```

```python
# Now we will do a naive linear regression
from sklearn.metrics import r2_score

model = linear_model.LinearRegression()

model.fit(X2_train, y2_train)
naive_preds = model.predict(X2_test)
r2_score(y2_test, naive_preds)
```

```python
# Do a linear regression after a basic split...

model1 = linear_model.LinearRegression()
model2 = linear_model.LinearRegression()

X2_test["ChocolateType"] = logistic_model.predict(X2_test[X1_test.drop(columns="SerialNo").columns])
X2_train["ChocolateType"] = logistic_model.predict(X2_train[X1_test.drop(columns="SerialNo").columns])


X2_test_gob = X2_test[X2_test["ChocolateType"] == "Gopstopper"]
X2_train_gob = X2_train[X2_train["ChocolateType"] == "Gopstopper"]


X2_test_nogob = X2_test[X2_test["ChocolateType"] != "Gopstopper" ]
X2_train_nogob = X2_train[X2_train["ChocolateType"] != "Gopstopper"]

y2_test_gop = y2_test[X2_test["ChocolateType"] == "Gopstopper"]
y2_test_nogop = y2_test[X2_test["ChocolateType"] != "Gopstopper"]

y2_train_gop = y2_train[X2_train["ChocolateType"] == "Gopstopper"]
y2_train_nogop = y2_train[X2_train["ChocolateType"] != "Gopstopper"]

model1.fit(X2_train_gob.drop(columns="ChocolateType"), y2_train_gop)
model2.fit(X2_train_nogob.drop(columns="ChocolateType"), y2_train_nogop)

y2_test_gop_prediction = model1.predict(X2_test_gob.drop(columns="ChocolateType"))
y2_test_nogop_prediction = model2.predict(X2_test_nogob.drop(columns="ChocolateType"))

print(r2_score(y2_test_gop, y2_test_gop_prediction))

print(r2_score(y2_test_nogop, y2_test_nogop_prediction))


y_full_pred = np.zeros(len(X2_test))
y_full_pred[X2_test["ChocolateType"] == "Gopstopper"] = y2_test_gop_prediction
y_full_pred[X2_test["ChocolateType"] != "Gopstopper"] = y2_test_nogop_prediction

print(r2_score(y2_test, y_full_pred))
```
